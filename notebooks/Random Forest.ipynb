## 01.Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../dataset/malicious_phish.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
#checking duplicates
df.duplicated('url').sum()
```

```python
df = df.drop_duplicates(subset='url')
df.shape
```

```python
print(df.duplicated('url').sum())
```

```python
#check class labels
df["type"].value_counts()
```

```python
#Standardize labels for binary classification
df["label_binary"] = df["type"].map({"benign": 0,"defacement": 1,"phishing": 1,"malware": 1})
#Standardize labels for multi-class classification
df["label_multi"] = df["type"].map({"benign": 0,"defacement": 1,"phishing": 2,"malware": 3})
```

```python
df.head()
```

```python
df["url"] = df["url"].str.lower()
```

```python
df["url"] = df["url"].str.strip()
```

```python
#clean url
df["clean_url"] = (
    df["url"]
    .str.replace(r"^https?://", "", regex=True)
    .str.replace(r"^www\.", "", regex=True)
)
```

```python
df.head()
```

```python
#choose features and labels for binary classification
X_binary = df["clean_url"]
y_binary = df["label_binary"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_binary, y_binary, test_size=0.2, random_state=42,stratify=y_binary)
```

```python
print(X_train.shape)
print(X_test.shape)

print(y_train.value_counts())
print(y_test.value_counts())
```

# `How do we convert URLs into numbers for Random Forest?`

- Random Forest cannot read text. It can only work with numbers.

```python
#Tf-idf vectorization
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
print(X_train_tfidf.shape)
print(X_test_tfidf.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
```

```python
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

What do these parameters mean?
- n_estimators=100
    - Build 100 decision trees.
- random_state=42
    - Makes results reproducible.
- n_jobs=-1
    - Uses all CPU cores to train faster.

```python
rf.fit(X_train_tfidf, y_train)
```

```python
y_pred = rf.predict(X_test_tfidf)
```

```python
y_prob = rf.predict_proba(X_test_tfidf)[:, 1]
```

```python
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
print("ROC AUC  :", roc_auc_score(y_test, y_prob))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
