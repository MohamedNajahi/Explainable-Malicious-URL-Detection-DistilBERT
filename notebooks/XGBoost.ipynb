```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../dataset/malicious_phish.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df = df.drop_duplicates(subset='url')
df.shape
```

```python
#checking duplicates
df.duplicated('url').sum()
```

```python
#check class labels
df["type"].value_counts()
```

```python
#Standardize labels for binary classification
df["label_binary"] = df["type"].map({"benign": 0,"defacement": 1,"phishing": 1,"malware": 1})
#Standardize labels for multi-class classification
df["label_multi"] = df["type"].map({"benign": 0,"defacement": 1,"phishing": 2,"malware": 3})
```

```python
df.head()
```

```python
df["url"] = df["url"].str.lower()
```

```python
df["url"] = df["url"].str.strip()
```

```python
#clean url
df["clean_url"] = (
    df["url"]
    .str.replace(r"^https?://", "", regex=True)
    .str.replace(r"^www\.", "", regex=True)
)
```
